# Train + Validate the Full Combined Agent (+ Conv1D) on Kaggle GPU

Clones `approach/full-combo-conv1d` and runs everything there. Forked
from `approach/full-combo` with one architectural addition: a small
Conv1D layer between the embedding and the LSTM, learning local
character-window patterns (a learned n-gram detector) before the
recurrence runs. English orthography is heavily locally-driven (q->u,
common digraphs, common endings), so this gives the LSTM a head start on
exactly the kind of pattern this task leans on most.

Four signals total, same as full-combo:

1. **Candidate-filtering** -- dictionary words still consistent with the
   board + wrong guesses, sharp once the pool narrows.
2. **Character n-gram fallback** -- CandidateAgent's own forward/backward
   n-gram model, reused directly (no dictionary match needed).
3. **Conv1D + BiLSTM** (attention + guessed-wrong/remaining features +
   cosine LR decay) -- the strongest neural variant built so far.
4. **Vowel-ratio guard** -- once over half the *revealed* letters are
   vowels, stop guessing further vowels.
